```python
import numpy as np
import pandas as pd
index=[('A',2010),('A',2020),('A',2025),('B',2010),('B',2020),('B',2025)]
population=[123,456,789,101,121,131]
i=pd.MultiIndex.from_tuples(index)
print(i)

pop=pd.Series(population,index=i)
print(pop)
print(pop['A',2020])
print(pop['A'])
print(pop[:,2010])
print(pop[pop>100])
pop_df=pop.unstack()
print(pop_df)
pop_df.stack()

```

```python
import pandas as pd
arrays = [
    ['A', 'A', 'A', 'B', 'B', 'B'],
    [2010, 2020, 2025, 2010, 2020, 2025]]

index = pd.MultiIndex.from_arrays(arrays)

print(index)
```

```python
import pandas as pd

index = pd.MultiIndex.from_product(
    [['A', 'B'], [2010, 2020, 2025]]
)

print(index)
```
